```python
import os

PATH = os.getcwd()

RESULTS_PATH = os.path.join(PATH, 'results')
EXPERIMENT_PATH = os.path.join(RESULTS_PATH, 'vgd_torch__megan_ensemble/debug')

print('path exists?', os.path.exists(EXPERIMENT_PATH))
print('is dir?', os.path.isdir(EXPERIMENT_PATH))
```

```python
from rich.pretty import pprint
from pycomex.functional.experiment import Experiment
from graph_attention_student.torch.megan import MeganEnsemble

experiment = Experiment.load(EXPERIMENT_PATH)
experiment.load(EXPERIMENT_PATH)

model_path = os.path.join(EXPERIMENT_PATH, 'ensemble.ckpt')
model: MeganEnsemble = MeganEnsemble.load(model_path)

print('experiment parameters:')
pprint(experiment.parameters, max_length=100)

print('experiment data:')
pprint(experiment.data, max_length=10)

print('model:')
print('strategy:', model.combine_importances_strategy)
```

```python
from visual_graph_datasets.web import ensure_dataset
from visual_graph_datasets.data import VisualGraphDatasetReader
from visual_graph_datasets.processing.molecules import MoleculeProcessing
from visual_graph_datasets.processing.colors import ColorProcessing
from visual_graph_datasets.util import dynamic_import

dataset_path = ensure_dataset('rb_dual_motifs')
reader = VisualGraphDatasetReader(dataset_path)
index_data_map = reader.read()
processing = ColorProcessing()

print('dataset processing:')
pprint(processing, max_length=10)

pprint(experiment.CHANNEL_INFOS)
```

```python
import tempfile
import matplotlib.pyplot as plt
import numpy as np
from visual_graph_datasets.visualization import draw_image
from visual_graph_datasets.visualization.importances import plot_edge_importances_background
from visual_graph_datasets.visualization.importances import plot_node_importances_background

plt.style.use('default')

INDEX = 22
NUM_CHANNELS = 2

model.combine_importances_strategy = 'mean'

metadata = index_data_map[INDEX]['metadata']

graph = metadata['graph']
image_path = index_data_map[INDEX]['image_path']
print('image exists?', os.path.exists(image_path))

result: dict = model.forward_graph(graph)

fig, rows = plt.subplots(nrows=2, ncols=NUM_CHANNELS, figsize=(16, 16), squeeze=False)
fig.suptitle(f'INDEX: {INDEX}\n'
                f'Prediction:  {np.round(result["graph_output"], 2)}\n'
                f'Uncertainty: {np.round(result["graph_uncertainty"], 2)}')

for k in range(2):
            
    ax_im = rows[0, k]
    ax_im.set_title(f'channel {k} - explanation')
    draw_image(ax=ax_im, image_path=image_path)
    plot_node_importances_background(
        ax=ax_im,
        g=graph,
        node_positions=graph['node_positions'],
        node_importances=result['node_importance'][:, k],
        v_max=np.max(result['node_importance']),
        v_min=0.0,
        color=experiment.CHANNEL_INFOS[str(k)]['color'],
    )
    plot_edge_importances_background(
        ax=ax_im,
        g=graph,
        node_positions=graph['node_positions'],
        edge_importances=result['edge_importance'][:, k],
        v_max=np.max(result['edge_importance']),
        v_min=0.0,
        color=experiment.CHANNEL_INFOS[str(k)]['color'],
    )
    
    ax_unc = rows[1, k]
    ax_unc.set_title(f'channel {k} - uncertainty')
    draw_image(ax=ax_unc, image_path=image_path)
    plot_node_importances_background(
        ax=ax_unc,
        g=graph,
        node_positions=graph['node_positions'],
        node_importances=result['node_importance_std'][:, k],
        color='purple',
    )
    plot_edge_importances_background(
        ax=ax_unc,
        g=graph,
        node_positions=graph['node_positions'],
        edge_importances=result['edge_importance_std'][:, k],
        color='purple',
    )

    fig.show()
    
```
